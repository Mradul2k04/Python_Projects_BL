 Correlation Analysis
 
Compute correlation matrix.


Plot heatmap using seaborn.


Identify strongest relationships.


```python
df.info()
```

```python
#Compute correlation matrix.
numeric_val=df.select_dtypes('float64')
corr_matrix=numeric_val.corr()
corr_matrix
```

```python
#Plot heatmap using seaborn
plt.figure(figsize=(10,5))
sns.heatmap(corr_matrix,annot=True,linewidths=0.5)
plt.title('Correlation Matrix of Numeric Features', fontsize=14,fontweight='bold')
plt.show()
```

```python
#Identify strongest relationships

corr_pairs = corr_matrix.unstack()
sorted_pairs = corr_pairs.sort_values(ascending=False)
unique_pairs = sorted_pairs[sorted_pairs < 1.0].drop_duplicates()
print(" STRONGEST POSITIVE RELATIONSHIPS (Both go up together):")
print(unique_pairs.head(3))

print("\n STRONGEST NEGATIVE RELATIONSHIPS (One goes up, the other goes down):")
print(unique_pairs.tail(3))
```

 Q11.Outlier Detection
 
Detect outliers using:


IQR OR Z-score (numpy)


Visualize using boxplot (seaborn).


Mention number of outliers found.


```python
#Detect outliers using:
#IQR

Q1 = df['Overall_Score'].quantile(0.25)
Q3 = df['Overall_Score'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
    
outliers = df[(df['Overall_Score'] < lower_bound) | (df['Overall_Score'] > upper_bound)]
print(f'Number of outliers found based on Overall_Score: {len(outliers)}')
```

```python
#Visualize using boxplot (seaborn).
plt.Figure(figsize=(9,6))
sns.boxplot(df,x="Overall_Score")
plt.title("Boxplot of Overall Score for Outlier Detection",fontsize=15,fontweight="bold")
plt.xlabel("Overall_Performance_Score")
plt.show()
```
