Q2. Data Cleaning

```python
df['RANK_2025']=df['RANK_2025'].astype(str).str.extract('(\d+)').astype(float)
```

```python
df['RANK_2024']=df['RANK_2024'].astype(str).str.extract('(\d+)').astype(float)
```

```python
df['Overall_Score']=pd.to_numeric(df['Overall_Score'],errors='coerce')
```

```python
df['Sustainability_Rank']=df['Sustainability_Rank'].str.replace(r'[+|=]','',regex=True)
df['Sustainability_Rank']=pd.to_numeric(df['Sustainability_Rank'],errors='coerce')
```

```python
df.info()
```

```python
df.isna()
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

 Feature Engineering (Using NumPy)
 
Create a new column Score_Category:


Elite → ≥ 90


High → 75–89


Medium → 50–74


Low → < 50


Use numpy.where() or vectorized operations.




```python
#Create a new column Score_Category
#Used numpy.where()
df['Score_Catrgory']=np.where(df['Overall_Score']>=90,'Elite',
                            np.where(df['Overall_Score']>=75,'High', 
                            np.where(df['Overall_Score']>=50,'Medium',"Low")))
df
```

```python
top_10_universities=df.sort_values('RANK_2025').head(10)[['RANK_2025','Institution_Name']]
top_10_universities
```

```python
plt.figure(figsize=(10, 8))
sns.barplot(x='Institution_Name',y='RANK_2025' ,data=top_10_universities)
plt.title('Top 10  Universities', fontsize=15, fontweight='bold')
plt.xticks(rotation=30, ha='right')
plt.xlabel('Institution_Name',fontsize=10, fontweight='bold')
plt.ylabel("RANK_2025",fontsize=10, fontweight='bold')
plt.show()
```
